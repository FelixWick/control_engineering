```python
import control as ct
import matplotlib.pyplot as plt
```

```python
s = ct.TransferFunction.s

G_inner = 1 / (0.5*s + 1)
G_outer = 1 / (5*s + 1)
```

```python
# Inner PI controller
C_inner = 2 + 5/s
T_inner = ct.feedback(C_inner * G_inner)
```

```python
# Outer PI controller
C_outer = 1 + 0.5/s
```

```python
# Full cascade system
T_total = ct.feedback(C_outer * G_outer * T_inner)
```

```python
t, y = ct.step_response(T_total)

plt.plot(t, y)
plt.title("Cascade Control")
plt.grid()
```
