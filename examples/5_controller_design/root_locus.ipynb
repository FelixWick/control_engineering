```python
import control as ct
import matplotlib.pyplot as plt
```

```python
s = ct.TransferFunction.s
G = 1 / (s * (s + 3) * (s + 5))
```

Root locus shows pole trajectories:

```python
ct.root_locus(G, grid=True)
```

The root locus shows where the closed-loop poles go as gain K increases. And therefore how stability and dynamics change.

Stability Information:
* Left half-plane (Re < 0) → stable
* Right half-plane (Re > 0) → unstable

So:
* If all branches stay left → stable for all gains
* If a branch crosses into right → instability occurs

#### Imaginary Axis Crossing

Where the locus crosses the imaginary axis: That gain value is the stability limit
* Before crossing → stable
* After crossing → unstable

#### Speed of Response

Pole location determines dynamics:
* Further left → faster system
* Closer to imaginary axis → slower
* Large imaginary part → oscillations

#### Damping and Oscillation

Complex poles:
* sigma → decay rate
* omega → oscillation frequency

Root locus shows:
* Whether oscillations appear
* How damped they are

#### Design use

1. Pick a gain K
2. Ensure poles are:
    * In left half-plane
    * At desired speed (far enough left)
    * With acceptable damping

#### Here:

* 3 poles
* No zeros

That means:
* 3 branches (one per pole)
* All branches must go to infinity (since no zeros)

1) Start (K = 0): Poles at 0, -3, -5

2) Increase K

* The poles at 0 and -3 move toward each other
* They meet somewhere between them → this is called a breakaway point

3) After meeting

* The two poles leave the real axis
* They become a complex conjugate pair
* They move:
    * upward (positive imaginary)
    * downward (negative imaginary)

4) Third pole

* The pole at -5 moves left toward −∞
* It stays on the real axis

```python
def plot_step_response(K):
  T = ct.feedback(K * G)
  t, y = ct.step_response(T)
  plt.figure()
  plt.plot(t, y)
  plt.title(f'Step Response with K = {K}')
  plt.xlabel('Time (s)')
  plt.ylabel('Output')
  plt.grid()
```

```python
for K in [1, 10, 100, 200]:
  plot_step_response(K)
```
