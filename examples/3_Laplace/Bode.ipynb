```python
import control as ct
```

```python
s = ct.TransferFunction.s
```

```python
# G = ct.tf([10], [0.1, 1, 0])
G = 10 / (s * (1 + 0.1 * s))
```

```python
ct.bode_plot(G, dB=True, Hz=False, grid=True, title="Bode Plot of G(s)", plot_magnitude=True, plot_phase=True)
```

```python
# G = ct.tf([0.14, 1.4], [0.01, 1.01, 1])
G = 1.4 * (1 + 0.1 * s) / ((1 + s) * (1 + 0.01 * s))
```

```python
ct.bode_plot(G, dB=True, Hz=False, grid=True, title="Bode Plot of G(s)")
```

1. Start is flat (0 dB/decade)

Interpretation:
	•	number of poles = number of zeros
	•	no integrator

2. First change at ω = 1
	•	slope: 0 → −20 dB/decade

Meaning: pole added

1 / (1 + s)

3. Second change at ω = 10
	•	slope: −20 → 0 dB/decade

Meaning: zero added

(1 + 0,1 s)

4. Third change at ω = 100
	•	slope: 0 → −20 dB/decade

Meaning: one more pole

1 / (1 + 0.01 s)

5. Combine (without Gain)

G(s) = K * (1 + 0.1 s) / ((1 + s)(1 + 0.01 s))

6. Determine gain

At  ω = 1: 0 dB

→ |G(j1)| = 1

- (1 + j) ≈ sqrt(2)
- (1 + j0.1) ≈ 1
- (1 + j0.01) ≈ 1

G(j1) ≈ K * 1 / (sqrt(2) * 1)

For G(j1) ≈ 1:

K ≈ sqrt(2) * 1.4

→ G(s) = 1.4 * (1 + 0.1 s) / ((1 + s)(1 + 0.01 s))

### Here is an example to see the importance of the phase plot:

System 1: two identical poles

```python
# G = ct.tf([1], [1, 2, 1])
G = 1 / (1 + s)**2
```

```python
ct.bode_plot(G, dB=True, Hz=False, grid=True, title="Bode Plot of G(s)")
```

System 2: two different poles

```python
# G = ct.tf([1], [0.1, 1.1, 1])
G = 1 / ((1 + s) * (1 + 0.1 * s))
```

```python
ct.bode_plot(G, dB=True, Hz=False, grid=True, title="Bode Plot of G(s)")
```

Phase plot:

- fast transition → identical poles
- slow transition → different poles

### PT2 example with resonance:

```python
# G = ct.tf([1], [1, 0.5, 1])
G = 1 / (s**2 + 0.5 * s + 1)
```

```python
ct.bode_plot(G, dB=True, Hz=False, grid=True, title="Bode Plot of G(s)")
```
