```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Continuous-time poles
s_poles_LHP = [-1, -2]    # stable
s_poles_RHP = [1, 2]      # unstable
```

```python
T = 0.1  # sampling time

z_LHP = [np.exp(p*T) for p in s_poles_LHP]
z_RHP = [np.exp(p*T) for p in s_poles_RHP]

print("LHP mapped to z-plane:", z_LHP)
print("RHP mapped to z-plane:", z_RHP)
```

```python
fig, ax = plt.subplots(figsize=(12,5))
for spine in ax.spines.values():
    spine.set_visible(False)
ax.set_xticks([])
ax.set_yticks([])

# --- s-plane subplot ---
ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(np.real(s_poles_LHP), np.imag(s_poles_LHP), 'bo', label='LHP (stable)')
ax1.plot(np.real(s_poles_RHP), np.imag(s_poles_RHP), 'rx', label='RHP (unstable)')
ax1.axhline(0, color='gray')
ax1.axvline(0, color='gray')
ax1.set_xlabel('Re(s)')
ax1.set_ylabel('Im(s)')
ax1.set_title('Continuous-Time s-plane')
ax1.grid(True)
ax1.legend()
ax1.set_aspect('equal', 'box')
ax1.set_xlim([-2.5, 2.5])
ax1.set_ylim([-2.5, 2.5])

# --- z-plane subplot ---
ax2 = fig.add_subplot(1, 2, 2)
# unit circle
theta = np.linspace(0, 2*np.pi, 200)
ax2.plot(np.cos(theta), np.sin(theta), 'k--', label='Unit Circle')
ax2.plot(np.real(z_LHP), np.imag(z_LHP), 'bo', markersize=8, label='LHP → stable')
ax2.plot(np.real(z_RHP), np.imag(z_RHP), 'rx', markersize=8, label='RHP → unstable')
ax2.axhline(0, color='gray')
ax2.axvline(0, color='gray')
ax2.set_xlabel('Re(z)')
ax2.set_ylabel('Im(z)')
ax2.set_title('Discrete-Time z-plane')
ax2.legend()
ax2.grid(True)
ax2.set_aspect('equal', 'box')
ax2.set_xlim([-1.5, 1.5])
ax2.set_ylim([-1.5, 1.5])

plt.tight_layout()
plt.show()
```
