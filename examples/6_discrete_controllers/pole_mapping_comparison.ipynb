```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sampling time
T = 1.0
```

```python
# Continuous-time pole
s = -1
```

```python
# Discretization mappings
z_zoh = np.exp(s * T)                     # ZOH (exact)
z_euler = 1 + s * T                       # Forward Euler
z_backward = 1 / (1 - s * T)              # Backward Euler
z_tustin = (1 + s*T/2) / (1 - s*T/2)      # Tustin
```

```python
# Print values (nice for debugging / lecture)
print("ZOH:", z_zoh)
print("Forward Euler:", z_euler)
print("Backward Euler:", z_backward)
print("Tustin:", z_tustin)
```

```python
fig, ax = plt.subplots(figsize=(6,6))

# Unit circle
theta = np.linspace(0, 2*np.pi, 300)
ax.plot(np.cos(theta), np.sin(theta), 'k--', linewidth=2)

# Plot poles
ax.plot(np.real(z_zoh), np.imag(z_zoh), 'go', label='ZOH (exact)', markersize=10)
ax.plot(np.real(z_euler), np.imag(z_euler), 'ro', label='Forward Euler', markersize=10)
ax.plot(np.real(z_backward), np.imag(z_backward), 'bo', label='Backward Euler', markersize=10)
ax.plot(np.real(z_tustin), np.imag(z_tustin), 'mo', label='Tustin', markersize=10)

# Axes
ax.axhline(0)
ax.axvline(0)

# Formatting
ax.set_aspect('equal', 'box')
ax.set_xlim([-1.5, 1.5])
ax.set_ylim([-1.5, 1.5])
ax.set_xlabel('Re(z)')
ax.set_ylabel('Im(z)')
ax.set_title('Discretization Methods: Pole Mapping to z-plane')
ax.legend()
ax.grid(True)

plt.show()
```
