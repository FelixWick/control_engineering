```python
import matplotlib.pyplot as plt
import control as ct
```

```python
G_s = ct.tf([1], [1, 1])
T = 1.0
```

```python
G_z = ct.c2d(G_s, T, method='zoh')
print(G_z)
```

```python
plt.figure()
t, y = ct.step_response(G_z)
# t, y = ct.step_response(ct.sample_system(G_s, T, method="zoh"))
plt.step(t, y, where="post", label="Discrete-time")

t, y = ct.step_response(G_s)
plt.plot(t, y, label="Continuous-time")

plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure()
for T in [0.1, 0.5, 1.0]:
    Gd = ct.sample_system(G_s, T, method="zoh")
    t, y = ct.step_response(Gd)
    plt.step(t, y, where="post", label=f"T = {T}")

plt.legend()
plt.grid()
plt.show()
```
